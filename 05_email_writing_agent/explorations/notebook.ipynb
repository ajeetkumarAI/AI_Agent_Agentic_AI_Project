## 2. Prepare and validate the request

```python
def validate_inputs(context: str, tone: str, recipient: str) -> tuple[str, str, str]:
    """Trim and validate the three inputs needed for an email."""

    values = {
        "context": context.strip(),
        "tone": tone.strip(),
        "recipient": recipient.strip(),
    }
    for name, value in values.items():
        if not value:
            raise ValueError(f"{name.capitalize()} cannot be empty.")
    if len(values["context"]) > MAX_CONTEXT_LENGTH:
        raise ValueError("Context is too long.")
    return values["context"], values["tone"], values["recipient"]


context = "Follow up on our product demo from last Tuesday. The client showed interest but has not responded."
tone = "professional and friendly"
recipient = "a potential client"
clean_context, clean_tone, clean_recipient = validate_inputs(context, tone, recipient)
print(clean_context)
print(clean_tone)
print(clean_recipient)
```
